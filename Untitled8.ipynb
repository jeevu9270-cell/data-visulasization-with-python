```python
import matplotlib.pyplot as plt
import pandas as pd

df = pd.read_csv("stock_data.csv", parse_dates=["Date"])

plt.figure(figsize=(12, 6))
plt.plot(df["Date"], df["Close"], color='green', marker='*', linewidth=2, markersize=4)

plt.title("closing price trend over time", fontsize=14, fontweight='bold')
plt.xlabel("Date")
plt.ylabel("closing price")
plt.grid(True, linestyle='--', alpha=0.5)
plt.xticks(rotation=45)
plt.tight_layout()
plt.savefig('line_plot.png', dpi=300)
plt.show()
```

```python
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns


df = pd.read_csv("car_data.csv")


sns.set_style("whitegrid")
sns.scatterplot(
    data=df,
    x="Engine_Capacity_cc",
    y="Fuel_Efficiency_km_l",
    s=70,
    color="green",
    edgecolor="black"
)

plt.title("Engine Capacity vs Fuel Efficiency")
plt.xlabel("Engine Capacity (cc)")
plt.ylabel("Fuel Efficiency (km/l)")
plt.tight_layout()
plt.savefig("scatter_plot.png", dpi=300)
plt.show()
```

```python
import matplotlib.pyplot as plt
import pandas as pd

df_patients = pd.read_csv("patient_data.csv", parse_dates=["Month"])

fig, (axes1, axes2) = plt.subplots(1, 2, figsize=(14, 5))


axes1.plot(df_patients["Month"], df_patients["Admitted"], marker='^', color='red')
axes1.set_title("Monthly Patient Admissions")
axes1.set_xlabel("Month")
axes1.set_ylabel("Patients Admitted")

axes2.bar(df_patients["Month"], df_patients["Emergency"], width=20, color="green")
axes2.set_title("Emergency Cases by Month")
axes2.set_xlabel("Month")
axes2.set_ylabel("Emergency Cases")

plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd

# Read data from CSV file
data = pd.read_csv("product_data.csv")

sns.set_theme(style="whitegrid")

fig, axes = plt.subplots(nrows=1, ncols=2, figsize=(12, 5))

# Updated x to use "Product_Category" with an underscore
sns.boxplot(data=data, x="Product_Category", y="Price", palette=['red', 'blue', 'green', 'orange'], ax=axes[0], hue="Product_Category")
axes[0].set_title("Product Price Distribution (Boxplot)")
axes[0].set_xlabel("Product Category")
axes[0].set_ylabel("Price")

# Updated x to use "Product_Category" with an underscore
sns.violinplot(data=data, x="Product_Category", y="Price", palette=['red', 'blue', 'green', 'orange'], ax=axes[1], hue="Product_Category")
axes[1].set_title("Product Price Distribution (Violinplot)")
axes[1].set_xlabel("Product Category")
axes[1].set_ylabel("Price")

fig.suptitle("Product Price Analysis")
plt.tight_layout()
plt.show()
```
